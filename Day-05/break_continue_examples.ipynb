```python
# Stop the loop when the value is found
for num in range(1, 10):
    if num == 5:
        print("Number 5 found, stopping the loop.")
        break
    print(num)
```

```python
# Skip even numbers
for num in range(1, 10):
    if num % 2 == 0:
        continue
    print(num)  # Prints only odd numbers
```
