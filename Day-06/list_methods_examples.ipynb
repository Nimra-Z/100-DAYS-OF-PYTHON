```python
# List Methods Examples

fruits = ["apple", "banana", "cherry"]

# Append a new item
fruits.append("orange")
print(fruits)  # ['apple', 'banana', 'cherry', 'orange']
```

```python
# Remove an item
fruits.remove("banana")
print(fruits)  # ['apple', 'cherry', 'orange']
```

```python
# Sort the list
fruits.sort()
print(fruits)  # ['apple', 'cherry', 'orange']
```

```python
# Reverse the list
fruits.reverse()
print(fruits)  # ['orange', 'cherry', 'apple']
```
