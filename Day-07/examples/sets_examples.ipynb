## Sets

Sets are unordered collections of unique elements in Python. They are useful for operations like union, intersection, and difference.

Key Features of Sets:

1. Mutable: Elements can be added or removed.
2. Unordered: No guarantee of the order of elements.
3. Unique: Duplicate elements are automatically removed.

```python
# Creating a set
empty_set = set()  # Note: {} creates an empty dictionary
my_set = {1, 2, 3, 4, 5}

# Mixed data types
mixed_set = {1, "Python", 3.14}

# Adding elements to a set
my_set.add(6)
print(my_set)  # Output: {1, 2, 3, 4, 5, 6}
```

```python
# Removing elements
my_set.remove(3)  # Raises KeyError if element not found
my_set.discard(10)  # Does not raise an error if element not found

# Checking membership
print(4 in my_set)  # Output: True

# Length of a set
print(len(my_set))  # Output: 5
```

```python
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}

# Union
print(set_a | set_b)  # Output: {1, 2, 3, 4, 5, 6}

# Intersection
print(set_a & set_b)  # Output: {3, 4}

# Difference
print(set_a - set_b)  # Output: {1, 2}

# Symmetric Difference
print(set_a ^ set_b)  # Output: {1, 2, 5, 6}
```

```python
# Sets: Basic Operations
set1 = {1, 2, 3}
set2 = {3, 4, 5}

# Union
print("Union:", set1 | set2)

# Intersection
print("Intersection:", set1 & set2)

# Difference
print("Difference (set1 - set2):", set1 - set2)

# Symmetric Difference
print("Symmetric Difference:", set1 ^ set2)

# Adding and removing elements
set1.add(6)
print("After adding 6:", set1)

set1.discard(2)
print("After discarding 2:", set1)
```
