## Tuples

Tuples are immutable sequences in Python, meaning they cannot be modified after creation. They are often used for fixed collections of items.

Key Features of Tuples:

1. Immutable: Once defined, the values cannot be changed.
2. Ordered: Items in a tuple maintain the order in which they are defined.
3. Allows Duplicates: Tuples can store duplicate values.

```python
# Creating a tuple
my_tuple = (1, 2, 3, 4)
print(my_tuple)

# Tuple with one element (comma is required)
single_element_tuple = (5,)
print(single_element_tuple)

# Empty tuple
empty_tuple = ()
print(empty_tuple)
```

```python
# Indexing
print(my_tuple[1])  # Output: 2

# Slicing
print(my_tuple[1:3])  # Output: (2, 3)

# Negative indexing
print(my_tuple[-1])  # Output: 4
```

```python
# Length of a tuple
print(len(my_tuple))

# Concatenation
new_tuple = my_tuple + (5, 6)
print(new_tuple)

# Repetition
print(my_tuple * 2)

# Membership
print(3 in my_tuple)  # Output: True
```

## Tuple Methods:

Tuples have two built-in methods:

• count(): Returns the number of occurrences of a value.

• index(): Returns the first index of a value.

```python
example_tuple = (1, 2, 3, 1, 2, 1)
print(example_tuple.count(1))  # Output: 3
print(example_tuple.index(3))  # Output: 2
```

```python
# Creating and accessing tuples
my_tuple = (10, 20, 30, 40)
print("Tuple:", my_tuple)
print("First Element:", my_tuple[0])
print("Slice:", my_tuple[1:3])

# Common Operations
new_tuple = my_tuple + (50, 60)
print("Concatenated Tuple:", new_tuple)

# Tuple Methods
print("Count of 20:", my_tuple.count(20))
print("Index of 30:", my_tuple.index(30))

# Immutability demonstration
try:
    my_tuple[0] = 100  # This will raise an error
except TypeError as e:
    print("Error:", e)
```
