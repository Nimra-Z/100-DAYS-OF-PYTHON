### Introduction to Dictionaries

Dictionaries in Python are a collection of key-value pairs. They are unordered, mutable, and indexed by keys (which must be immutable types like strings, numbers, or tuples).


#### Creating a Dictionary:

```python
# Creating a Dictionary:
person = {
    "name": "Alice",
    "age": 25,
    "city": "New York"
}
print(person)
```

#### Accessing Values:

```python
# Accessing using keys
print(person["name"])  # Output: Alice

# Using `.get()` method (avoids KeyError if the key doesn't exist)
print(person.get("age"))  # Output: 25
print(person.get("gender", "Not specified"))  # Output: Not specified
```

#### Adding/Updating Items:

```python
# Add new key-value pair
person["gender"] = "Female"

# Update existing key
person["age"] = 26
print(person)
```

#### Removing Items:

```python
# Using `del`
del person["city"]

# Using `.pop()`
removed_value = person.pop("gender")
print(person, removed_value)

# Clear the entire dictionary
person.clear()
print(person)  # Output: {}
```

#### Dictionary Methods:

```python
person = {"name": "Alice", "age": 25, "city": "New York"}

# Getting all keys
print(person.keys())  # Output: dict_keys(['name', 'age', 'city'])

# Getting all values
print(person.values())  # Output: dict_values(['Alice', 25, 'New York'])

# Getting all key-value pairs
print(person.items())  # Output: dict_items([('name', 'Alice'), ('age', 25), ('city', 'New York')])

# Check if a key exists
print("name" in person)  # Output: True
print("gender" in person)  # Output: False
```

#### Iterating Over a Dictionary:

```python
for key, value in person.items():
    print(f"{key}: {value}")
```
