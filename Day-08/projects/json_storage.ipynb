```python
# JSON-like Dictionary Storage

import json

def save_to_file(dictionary, filename):
    with open(filename, 'w') as file:
        json.dump(dictionary, file)
    print(f"Dictionary saved to {filename}")

def load_from_file(filename):
    try:
        with open(filename, 'r') as file:
            return json.load(file)
    except FileNotFoundError:
        print(f"{filename} not found. Returning empty dictionary.")
        return {}

if __name__ == "__main__":
    dictionary = load_from_file("data.json")

    while True:
        print("\nDictionary Storage")
        print("1. Add/Update Key-Value Pair")
        print("2. View Dictionary")
        print("3. Delete Key")
        print("4. Save and Exit")

        choice = int(input("Enter your choice: "))
        if choice == 1:
            key = input("Enter key: ")
            value = input("Enter value: ")
            dictionary[key] = value
            print("Key-value pair added/updated.")
        elif choice == 2:
            print("Current Dictionary:", dictionary)
        elif choice == 3:
            key = input("Enter key to delete: ")
            if key in dictionary:
                del dictionary[key]
                print(f"Key '{key}' deleted.")
            else:
                print(f"Key '{key}' not found.")
        elif choice == 4:
            save_to_file(dictionary, "data.json")
            print("Exiting...")
            break
        else:
            print("Invalid choice. Try again.")
```
