```python
# Dictionary Manager

dictionary = {}

while True:
    print("\nDictionary Manager")
    print("1. Add key-value pair")
    print("2. View dictionary")
    print("3. Update key-value")
    print("4. Delete key-value")
    print("5. Exit")
    
    choice = int(input("Enter your choice: "))
    
    if choice == 1:
        key = input("Enter key: ")
        value = input("Enter value: ")
        dictionary[key] = value
        print("Key-value pair added!")
    elif choice == 2:
        print("Current Dictionary:", dictionary)
    elif choice == 3:
        key = input("Enter key to update: ")
        if key in dictionary:
            value = input("Enter new value: ")
            dictionary[key] = value
            print("Key-value pair updated!")
        else:
            print("Key not found!")
    elif choice == 4:
        key = input("Enter key to delete: ")
        if key in dictionary:
            del dictionary[key]
            print("Key-value pair deleted!")
        else:
            print("Key not found!")
    elif choice == 5:
        print("Exiting...")
        break
    else:
        print("Invalid choice. Try again.")
```
