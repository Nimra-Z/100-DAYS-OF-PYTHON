### Positional vs Keyword Arguments
Positional arguments are passed in the order they are defined.

Keyword arguments allow you to explicitly specify parameter names.

```python
# Example: Positional vs Keyword Arguments

def describe_pet(name, species):
    """Describes a pet using its name and species."""
    print(f"{name} is a {species}.")

# Using positional arguments
describe_pet("Buddy", "dog")

# Using keyword arguments
describe_pet(species="cat", name="Whiskers")
```
