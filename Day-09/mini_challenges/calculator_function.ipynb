```python
# Challenge: Calculator Function

def calculator(a, b, operator):
    """Performs a calculation based on the operator."""
    if operator == '+':
        return a + b
    elif operator == '-':
        return a - b
    elif operator == '*':
        return a * b
    elif operator == '/':
        return a / b
    else:
        return "Invalid operator"

# Test the function
print(calculator(10, 5, '+'))  # Output: 15
print(calculator(10, 5, '/'))  # Output: 2.0
```
