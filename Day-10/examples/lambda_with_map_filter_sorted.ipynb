```python
# Using lambda with map, filter, and sorted

numbers = [1, 2, 3, 4, 5]

# Using map to square numbers
squared = list(map(lambda x: x ** 2, numbers))
print(squared)  # Output: [1, 4, 9, 16, 25]

# Using filter to select odd numbers
odds = list(filter(lambda x: x % 2 != 0, numbers))
print(odds)  # Output: [1, 3, 5]

# Using sorted to order by custom key
sorted_desc = sorted(numbers, key=lambda x: -x)
print(sorted_desc)  # Output: [5, 4, 3, 2, 1]
```
