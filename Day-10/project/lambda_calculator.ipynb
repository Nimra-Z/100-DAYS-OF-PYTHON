```python
# Project: Lambda Calculator

# Define lambda functions for each operation
add = lambda a, b: a + b
subtract = lambda a, b: a - b
multiply = lambda a, b: a * b
divide = lambda a, b: a / b if b != 0 else "Error: Division by zero"

# Test the calculator
print("Add:", add(10, 5))        # Output: 15
print("Subtract:", subtract(10, 5))  # Output: 5
print("Multiply:", multiply(10, 5))  # Output: 50
print("Divide:", divide(10, 5))      # Output: 2.0
print("Divide by Zero:", divide(10, 0))  # Output: Error: Division by zero
```
